```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegressionCV
from sklearn.linear_model import LogisticRegressionCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix, roc_auc_score)
```

```python
df = pd.read_csv("train.csv")
df = df.dropna()
```

```python
print(df.columns)
```

```python
df["high_pollution"] = (df["pm2_5_mean"] > 150).astype(float)
```

```python
X = df.drop(columns=["pm2_5_mean", "high_pollution"])
y = df["high_pollution"]
```

```python
print(X.isna().sum())
print(np.isinf(X.select_dtypes(include="number")).sum())
```

```python
split = int(len(df) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]
```

```python
X = X.select_dtypes(include="number")
X_train, X_test = X.iloc[:split], X.iloc[split:]
print(X_train.shape, y_train.value_counts())
```

```python
def build(pen):
    return make_pipeline(
        StandardScaler(),
        LogisticRegressionCV(
            Cs=np.logspace(-3, 2, 20),
            cv=5,
            penalty=pen,
            solver="saga",          # works for both l1 and l2
            scoring="roc_auc",
            max_iter=5000,
        ),
    )
```

```python
for pen in ["l2", "l1"]:
    model = build(pen)
    model.fit(X_train, y_train)
    clf = model[-1]

    proba = model.predict_proba(X_test)[:, 1]
    pred = (proba >= 0.5).astype(int)

    print(f"\n===== {pen.upper()} =====")
    print("best C:", clf.C_[0])
    print("test AUC:", roc_auc_score(y_test, proba))
    print(confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred))

    coefs = pd.Series(clf.coef_[0], index=X.columns).sort_values()
    print(coefs)
```
